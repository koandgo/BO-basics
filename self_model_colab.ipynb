```python
# ---- Fit PCA on the full kraken descriptor table and expose the loadings ----
# kraken_features_only.csv = one row per ligand ("id") + the 190 physical-organic descriptors.
# Per the SI, descriptors are standard-scaled BEFORE PCA.

df_desc = pd.read_csv(DATA_DIR / "kraken_features_only.csv")

# Descriptor columns = everything except the ligand identifier
DESC_ID_COL = "id"
desc_cols = [c for c in df_desc.columns if c != DESC_ID_COL]

# Drop any descriptor that is entirely NaN, then fill remaining gaps with column means
X_desc = df_desc[desc_cols].apply(pd.to_numeric, errors="coerce")
X_desc = X_desc.dropna(axis=1, how="all")
desc_cols = X_desc.columns.tolist()
X_desc = X_desc.fillna(X_desc.mean())

N_COMPONENTS = 10  # keep the first 10 PCs (paper reports variance for PC1..PC10)

pca_pipe = Pipeline([
    ("scale", StandardScaler()),
    ("pca", PCA(n_components=N_COMPONENTS, random_state=42)),
])
scores = pca_pipe.fit_transform(X_desc)                 # per-ligand PC scores
pca = pca_pipe.named_steps["pca"]

pc_names = [f"PC{i+1}" for i in range(N_COMPONENTS)]

# Loadings: rows = original descriptors, cols = PCs. (components_ is [n_pc, n_features])
loadings = pd.DataFrame(pca.components_.T, index=desc_cols, columns=pc_names)

# Explained variance (sanity check against Table S7: ~28.0, 12.9, 11.2, 6.6 ...)
explained = pd.Series(pca.explained_variance_ratio_, index=pc_names, name="explained_var_ratio")
print("Explained variance ratio:")
print((explained * 100).round(1).to_string())
loadings.head()
```

```python
# ---- Rank feature contributions for any PC ----
def rank_loadings(pc, n=None, by_abs=True):
    """Return descriptors ranked by their loading on `pc` (e.g. 'PC1')."""
    s = loadings[pc]
    order = s.abs().sort_values(ascending=False) if by_abs else s.sort_values(ascending=False)
    out = pd.DataFrame({"loading": s.loc[order.index], "abs_loading": s.loc[order.index].abs()})
    return out.head(n) if n else out

# Example: top 10 contributors to PC1
rank_loadings("PC1", 10)
```

```python
# ---- Top-3 descriptors from each of PC1..PC4 -> merge onto the interpretable / one-hot table ----
TOP_K = 3
TOP_PCS = ["PC1", "PC2", "PC3", "PC4"]

top_desc_per_pc = {pc: rank_loadings(pc, TOP_K).index.tolist() for pc in TOP_PCS}
for pc, cols in top_desc_per_pc.items():
    print(pc, "->", cols)

# De-duplicated, order-preserving list of the raw descriptor columns to add
pc_top_features = list(dict.fromkeys(c for cols in top_desc_per_pc.values() for c in cols))
print("\nDescriptors added from PCs:", pc_top_features)

# Per-ligand lookup table of just those descriptors, keyed by kraken id
desc_lookup = df_desc[[DESC_ID_COL] + pc_top_features].copy()

# Choose the base table to augment. Prefer df_onehot_ligand if you've already built it;
# otherwise fall back to df_interp. Both are keyed to a ligand via kraken id.
base_df = df_onehot_ligand.copy() if "df_onehot_ligand" in dir() else df_interp.copy()

# Figure out which column in base_df holds the kraken ligand id.
id_candidates = [c for c in ("kraken_id", "id", "ligand") if c in base_df.columns]
if not id_candidates:
    raise KeyError("No kraken id column ('kraken_id'/'id'/'ligand') found in base_df to merge on.")
BASE_ID_COL = id_candidates[0]

df_pca_augmented = base_df.merge(
    desc_lookup, how="left", left_on=BASE_ID_COL, right_on=DESC_ID_COL,
)
if DESC_ID_COL != BASE_ID_COL and DESC_ID_COL in df_pca_augmented.columns:
    df_pca_augmented = df_pca_augmented.drop(columns=[DESC_ID_COL])

missing = df_pca_augmented[pc_top_features].isna().any(axis=1).sum()
print(f"\nMerged on '{BASE_ID_COL}'. Rows with unmatched descriptors: {missing}")
print("df_pca_augmented shape:", df_pca_augmented.shape)
df_pca_augmented.head()
```

## GP yield models â€” functions, settings, then scoring cells

Three cells set this up, then each scoring scheme is a single function call:

1. **Functions cell** â€” `GPExperimentConfig` (the settings container), `build_Xy`/`make_gp`,
   the experiment-logging helpers (`save_model_results`, `save_model_object`), and the
   scoring functions: `run_lolo_single` / `run_lolo_pooled` / `run_lolo_by_ligand` (three
   LOLO variants, see below) plus `run_kfold` / `run_traintest`. Nothing here should need
   editing to change a hyperparameter or dataframe.

   The three LOLO functions all share `_fit_lolo_folds` (fits one train/test fold per
   ligand) and differ only in how they slice/aggregate its output:
   - `run_lolo_single(cfg)` â€” holds out exactly one ligand, `cfg.holdout_ligand`.
   - `run_lolo_pooled(cfg)` â€” folds over every ligand, pools every held-out prediction
     into one R^2/RMSE per dataframe (implicitly weights by row count).
   - `run_lolo_by_ligand(cfg)` â€” folds over every ligand, scores each ligand
     separately, then averages those per-ligand scores per dataframe (equal weight per
     ligand) and writes the averages to `lolo_ligand_averages.csv`.

   Each call to a scoring function allocates its own fresh results folder, so
   different scoring schemes never mix their `model_results.csv`/`model_features.csv`/plots,
   and re-running the same scoring cell again later the same day doesn't overwrite or
   append into the previous run's folder either:
   `<cfg.results_root>/<today>/<run>/`, where `<run>` is `lolo_single`, `lolo_pooled`,
   `lolo_by_ligand`, `<n_splits>fold`, or `traintest_<pct>_<pct>`. If that folder already
   exists (e.g. the cell was already run today), the new run instead gets
   `<run>_1`, `<run>_2`, etc. â€” whichever suffix is first unused.
2. **Settings cell** â€” the kernel, `GP_KWARGS`, the `DATAFRAMES` registry, ligand-feature
   rules, split settings, and logging flags, all bundled into a single `CFG` object. Edit
   this cell to change how the models run.
3. **Scoring cells** â€” each is just `results_x = run_x(CFG)`.

```python
# ===== GP MODEL FUNCTIONS (edit hyperparameters/settings in the next cell instead) =====
from dataclasses import dataclass, field, replace
from typing import Callable, Optional
from sklearn.model_selection import KFold, train_test_split

@dataclass
class GPExperimentConfig:
    dataframes: dict
    kernel_fn: Callable[[int], object]
    gp_kwargs: dict
    ligand_col: str = "ligand"
    ligand_feature_models: set = field(default_factory=lambda: {"df_onehot_ligand"})
    holdout_ligand: Optional[str] = None
    n_splits: int = 5
    kfold_random_state: int = 42
    test_size: float = 0.2
    split_random_state: int = 42
    results_root: Path = Path("data") / "GP_results"
    save_results: bool = True

# --- Build (X, y, groups) from a dataframe: one-hot categoricals, scale numerics -----
def build_Xy(df, name, cfg):
    """Return (preprocessor, X_frame, y, groups) for a given dataframe.

    `name` is the cfg.dataframes registry key -- used only to decide whether the
    "ligand" column is allowed to stay in the feature matrix (see cfg.ligand_feature_models).
    """
    d = df.copy()
    y = d[TARGET].to_numpy(dtype=float)

    # group labels for leave-one-ligand-out (fall back to None if absent)
    groups = d[cfg.ligand_col].to_numpy() if cfg.ligand_col in d.columns else None

    # feature frame = everything except target and obvious id/leakage columns
    drop_cols = [c for c in [TARGET, "kraken_id", "id"] if c in d.columns]
    if cfg.ligand_col in d.columns and name not in cfg.ligand_feature_models:
        drop_cols.append(cfg.ligand_col)
    X = d.drop(columns=drop_cols)

    # split numeric vs categorical (avoid select_dtypes(include=["object"]) which
    # triggers the pandas object/str deprecation warning -- everything non-numeric
    # is treated as categorical instead)
    num_cols = X.select_dtypes(include=[np.number]).columns.tolist()
    cat_cols = [c for c in X.columns if c not in num_cols]

    pre = ColumnTransformer([
        ("num", StandardScaler(), num_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), cat_cols),
    ])
    return pre, X, y, groups

def make_gp(n_features, cfg):
    return GaussianProcessRegressor(kernel=cfg.kernel_fn(n_features), **cfg.gp_kwargs)

# --- Experiment logging: results CSV + feature-list CSV + pred-vs-obs plot, one ------
#     freshly-named subfolder per scoring RUN (i.e. per call to run_lolo_single(cfg),
#     run_kfold(cfg), etc.), nested under one subfolder per calendar day, under
#     <cfg.results_root>/ ----------------------------------------------------------
def _next_available_dir(base_dir):
    """Return `base_dir`, creating it, unless it already exists -- in which case try
    `<base_dir>_1`, `<base_dir>_2`, ... and create+return the first name that's free.
    This means re-running the same scoring cell on the same day never mixes its
    results into a previous run's folder; each execution gets its own directory."""
    candidate = base_dir
    i = 1
    while candidate.exists():
        candidate = base_dir.with_name(f"{base_dir.name}_{i}")
        i += 1
    candidate.mkdir(parents=True)
    return candidate

def _new_run_dir(cfg, run):
    """Allocate a fresh results directory for one invocation of a run_* scoring
    function: <cfg.results_root>/<today>/<run>[_1, _2, ...]."""
    return _next_available_dir(cfg.results_root / date.today().isoformat() / run)

def save_model_results(model, label, gp, feature_names,
                        n_transformed_features_min, n_transformed_features_max,
                        n_transformed_features_last, n_fit_rows,
                        y_true, y_pred, y_std, fit_seconds, cfg, out_dir, save=None):
    """Log one model run: append a metrics row to model_results.csv, the feature list
    to model_features.csv, and save a predicted-vs-observed plot -- all under `out_dir`
    (one directory per run_* invocation, see _new_run_dir).

    `model` identifies the feature set (a cfg.dataframes key); `label` identifies the
    specific fold/split within that scoring run (e.g. "LOLO_<ligand>", "5fold",
    "traintest_80_20"). n_transformed_features_{min,max,last} let LOLO/K-fold report how
    the post-one-hot feature count varied across folds (a single-fit scheme just repeats
    the same value three times); fitted_kernel_last is always just the last fold/fit's
    kernel, since kernels don't have a meaningful min/max. Pass save=False (or set
    cfg.save_results=False) to skip writing to disk.
    """
    if save is None:
        save = cfg.save_results
    if not save:
        return None

    y_true = np.asarray(y_true); y_pred = np.asarray(y_pred); y_std = np.asarray(y_std)

    rmse = root_mean_squared_error(y_true, y_pred)
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    mean_predictive_std = float(np.mean(y_std))

    row = pd.DataFrame([{
        "model": model, "label": label,
        "n_transformed_features_min": n_transformed_features_min,
        "n_transformed_features_max": n_transformed_features_max,
        "n_transformed_features_last": n_transformed_features_last,
        "n_fit_rows": n_fit_rows,
        "fit_seconds": fit_seconds, "fitted_kernel_last": str(gp.kernel_),
        "rmse": rmse, "mae": mae, "r2": r2,
        "mean_predictive_std": mean_predictive_std,
    }])
    results_path = out_dir / "model_results.csv"
    row.to_csv(results_path, mode="a", header=not results_path.exists(), index=False)

    features_path = out_dir / "model_features.csv"
    feat_df = pd.DataFrame({"model": model, "feature": feature_names})
    if features_path.exists():
        feat_df = pd.concat([pd.read_csv(features_path), feat_df]).drop_duplicates()
    feat_df.to_csv(features_path, index=False)

    fig, ax = plt.subplots(figsize=(5, 5))
    ax.scatter(y_true, y_pred, alpha=0.5, s=15)
    lims = [min(y_true.min(), y_pred.min()), max(y_true.max(), y_pred.max())]
    ax.plot(lims, lims, "k--", linewidth=1)
    ax.set_xlabel("Observed yield")
    ax.set_ylabel("Predicted yield")
    ax.set_title(f"{model} | {label}\nR^2 = {r2:.3f}, RMSE = {rmse:.3f}")
    fig.tight_layout()
    fig.savefig(out_dir / f"{model}__{label}_pred_vs_obs.png", dpi=150)
    plt.close(fig)

    return row.iloc[0].to_dict()

# --- Optional: persist the fitted GP + its fitted preprocessor for later reuse ------
# NOTE: a fitted GaussianProcessRegressor stores an n x n Cholesky factor of the
# kernel matrix, so these files can be large (tens of MB per single-fit model here;
# much more if called once per LOLO/K-fold fold). Not called by default -- uncomment
# the save_model_object(...) line in a scoring cell to opt in for that cell.
def save_model_object(model, label, gp, pre, out_dir):
    """Persist {'gp': gp, 'preprocessor': pre} to
    <out_dir>/<model>__<label>_model.joblib (joblib.load(path) to reuse)."""
    path = out_dir / f"{model}__{label}_model.joblib"
    joblib.dump({"gp": gp, "preprocessor": pre}, path)
    return path

# --- Fit on one fold, predict its held-out rows. Shared by run_lolo*/run_kfold. ------
def _fit_predict_fold(pre, X, y, tr, te, cfg):
    Xtr = pre.fit_transform(X.iloc[tr]); Xte = pre.transform(X.iloc[te])
    gp = make_gp(Xtr.shape[1], cfg)
    t0 = time.perf_counter()
    gp.fit(Xtr, y[tr])
    fit_seconds = time.perf_counter() - t0
    mean, std = gp.predict(Xte, return_std=True)
    return gp, mean, std, fit_seconds, Xtr.shape[0], Xtr.shape[1]

# --- Shared LOLO fold-fitting: fits one fold per requested ligand, for every --------
#     dataframe in cfg.dataframes. The three run_lolo_* functions below all just
#     slice/aggregate this raw per-(model, ligand) output differently.
def _fit_lolo_folds(cfg, ligands=None):
    """Fit one leave-one-ligand-out fold per ligand in `ligands` (default: every
    ligand present) for each dataframe. Returns
        {model_name: {ligand: {"y_true", "y_pred", "y_std", "gp",
                                "X_columns", "n_feat", "fit_seconds", "n_rows"}}}
    """
    logo = LeaveOneGroupOut()
    out = {}

    for name, df in cfg.dataframes.items():
        pre, X, y, groups = build_Xy(df, name, cfg)
        if groups is None:
            print(f"{name}: no '{cfg.ligand_col}' column -> skipping leave-one-ligand-out")
            continue

        wanted = set(ligands) if ligands is not None else set(np.unique(groups))
        for lig in wanted - set(groups):
            print(f"{name}: ligand '{lig}' not found -> skipping")
        wanted &= set(groups)
        if not wanted:
            continue

        out[name] = {}
        for tr, te in logo.split(X, y, groups):
            lig = groups[te][0]
            if lig not in wanted:
                continue
            gp, mean, std, fs, n_rows, n_feat = _fit_predict_fold(pre, X, y, tr, te, cfg)
            out[name][lig] = {
                "y_true": y[te], "y_pred": mean, "y_std": std, "gp": gp,
                "X_columns": X.columns.tolist(), "n_feat": n_feat,
                "fit_seconds": fs, "n_rows": n_rows,
            }

    return out

# ===== SCORING SCHEMES ==================================================================

# --- LOLO model 1: hold out exactly one ligand (cfg.holdout_ligand) ------------------
def run_lolo_single(cfg):
    """Leave-one-ligand-out for a single, specific held-out ligand: cfg.holdout_ligand.
    Trains on every other ligand, scores only on the held-out one."""
    if cfg.holdout_ligand is None:
        raise ValueError("run_lolo_single requires cfg.holdout_ligand to be set to a ligand name")

    run = "lolo_single"
    out_dir = _new_run_dir(cfg, run) if cfg.save_results else None
    folds = _fit_lolo_folds(cfg, ligands=[cfg.holdout_ligand])
    label = f"LOLO_{cfg.holdout_ligand}"
    results = {}

    for name, per_ligand in folds.items():
        fold = per_ligand.get(cfg.holdout_ligand)
        if fold is None:
            continue

        r2 = r2_score(fold["y_true"], fold["y_pred"])
        rmse = root_mean_squared_error(fold["y_true"], fold["y_pred"])
        results[name] = {"r2": r2, "rmse": rmse}
        print(f"{name}: leave-one-ligand-out R^2 = {r2:.3f}, RMSE = {rmse:.3f}  "
              f"(held out: '{cfg.holdout_ligand}')")

        save_model_results(
            model=name, label=label, gp=fold["gp"], feature_names=fold["X_columns"],
            n_transformed_features_min=fold["n_feat"], n_transformed_features_max=fold["n_feat"],
            n_transformed_features_last=fold["n_feat"], n_fit_rows=fold["n_rows"],
            y_true=fold["y_true"], y_pred=fold["y_pred"], y_std=fold["y_std"],
            fit_seconds=fold["fit_seconds"], cfg=cfg, out_dir=out_dir,
        )
        # save_model_object(model=name, label=label, gp=fold["gp"], pre=pre, out_dir=out_dir)

    print(f"\nSummary (LOLO, held out '{cfg.holdout_ligand}'):",
          {k: {m: round(v, 3) for m, v in d.items()} for k, d in results.items()})
    return results

# --- LOLO model 2: fold over every ligand, pool all held-out predictions together ----
def run_lolo_pooled(cfg):
    """Leave-one-ligand-out folded over every ligand, with every held-out prediction
    pooled together into a single R^2/RMSE per dataframe (ignores cfg.holdout_ligand --
    always uses every ligand)."""
    run = "lolo_pooled"
    out_dir = _new_run_dir(cfg, run) if cfg.save_results else None
    folds = _fit_lolo_folds(cfg, ligands=None)
    results = {}

    for name, per_ligand in folds.items():
        if not per_ligand:
            continue

        y_true_all = np.concatenate([f["y_true"] for f in per_ligand.values()])
        y_pred_all = np.concatenate([f["y_pred"] for f in per_ligand.values()])
        y_std_all = np.concatenate([f["y_std"] for f in per_ligand.values()])
        n_feat_all = [f["n_feat"] for f in per_ligand.values()]
        fit_seconds = sum(f["fit_seconds"] for f in per_ligand.values())
        n_fit_rows = sum(f["n_rows"] for f in per_ligand.values())
        last_fold = list(per_ligand.values())[-1]

        r2 = r2_score(y_true_all, y_pred_all)
        rmse = root_mean_squared_error(y_true_all, y_pred_all)
        results[name] = {"r2": r2, "rmse": rmse}
        print(f"{name}: leave-one-ligand-out R^2 = {r2:.3f}, RMSE = {rmse:.3f}  "
              f"(held out: all {len(per_ligand)} ligands, pooled)")

        save_model_results(
            model=name, label="LOLO_all_pooled", gp=last_fold["gp"], feature_names=last_fold["X_columns"],
            n_transformed_features_min=min(n_feat_all), n_transformed_features_max=max(n_feat_all),
            n_transformed_features_last=n_feat_all[-1], n_fit_rows=n_fit_rows,
            y_true=y_true_all, y_pred=y_pred_all, y_std=y_std_all, fit_seconds=fit_seconds,
            cfg=cfg, out_dir=out_dir,
        )
        # gp/pre here are only the LAST fold's fit. Uncomment to persist it:
        # save_model_object(model=name, label="LOLO_all_pooled", gp=last_fold["gp"], pre=pre, out_dir=out_dir)

    print("\nSummary (LOLO, pooled across ligands):",
          {k: {m: round(v, 3) for m, v in d.items()} for k, d in results.items()})
    return results

# --- LOLO model 3: fold over every ligand, score each ligand separately, average -----
def run_lolo_by_ligand(cfg):
    """Leave-one-ligand-out folded over every ligand, scored (and logged) separately
    per held-out ligand, then averaged per dataframe (ignores cfg.holdout_ligand --
    always uses every ligand). Writes the per-model averages to
    lolo_ligand_averages.csv, in addition to the per-ligand rows in model_results.csv."""
    run = "lolo_by_ligand"
    out_dir = _new_run_dir(cfg, run) if cfg.save_results else None
    folds = _fit_lolo_folds(cfg, ligands=None)
    per_ligand_metrics = {name: {} for name in folds}

    for name, ligand_folds in folds.items():
        for lig, f in ligand_folds.items():
            r2 = r2_score(f["y_true"], f["y_pred"])
            rmse = root_mean_squared_error(f["y_true"], f["y_pred"])
            per_ligand_metrics[name][lig] = {"r2": r2, "rmse": rmse}

            save_model_results(
                model=name, label=f"LOLO_{lig}", gp=f["gp"], feature_names=f["X_columns"],
                n_transformed_features_min=f["n_feat"], n_transformed_features_max=f["n_feat"],
                n_transformed_features_last=f["n_feat"], n_fit_rows=f["n_rows"],
                y_true=f["y_true"], y_pred=f["y_pred"], y_std=f["y_std"],
                fit_seconds=f["fit_seconds"], cfg=cfg, out_dir=out_dir,
            )

    averages = {}
    rows = []
    for name, ligand_metrics in per_ligand_metrics.items():
        if not ligand_metrics:
            continue
        r2s = [m["r2"] for m in ligand_metrics.values()]
        rmses = [m["rmse"] for m in ligand_metrics.values()]
        averages[name] = {
            "mean_r2": float(np.mean(r2s)), "std_r2": float(np.std(r2s)),
            "mean_rmse": float(np.mean(rmses)), "std_rmse": float(np.std(rmses)),
            "n_ligands": len(ligand_metrics),
        }
        rows.append({"model": name, **averages[name]})

    print("\nSummary (LOLO, averaged over ligands):")
    for name, m in averages.items():
        print(f"  {name}: mean R^2 = {m['mean_r2']:.3f} +/- {m['std_r2']:.3f}, "
              f"mean RMSE = {m['mean_rmse']:.3f} +/- {m['std_rmse']:.3f}  (n={m['n_ligands']} ligands)")

    if cfg.save_results and rows:
        out_path = out_dir / "lolo_ligand_averages.csv"
        pd.DataFrame(rows).to_csv(out_path, index=False)
        print(f"\nSaved per-ligand LOLO averages to {out_path}")

    return averages

def run_kfold(cfg):
    """cfg.n_splits-fold cross-validation."""
    run = f"{cfg.n_splits}fold"
    out_dir = _new_run_dir(cfg, run) if cfg.save_results else None
    kf = KFold(n_splits=cfg.n_splits, shuffle=True, random_state=cfg.kfold_random_state)
    results = {}

    for name, df in cfg.dataframes.items():
        pre, X, y, _ = build_Xy(df, name, cfg)

        y_true_all, y_pred_all, y_std_all, n_feat_all = [], [], [], []
        fit_seconds, n_fit_rows = 0.0, 0
        for tr, te in kf.split(X):
            gp, mean, std, fs, n_rows, n_feat = _fit_predict_fold(pre, X, y, tr, te, cfg)
            y_pred_all.append(mean); y_std_all.append(std); y_true_all.append(y[te])
            fit_seconds += fs; n_fit_rows += n_rows; n_feat_all.append(n_feat)

        y_true_all = np.concatenate(y_true_all)
        y_pred_all = np.concatenate(y_pred_all)
        y_std_all = np.concatenate(y_std_all)

        r2 = r2_score(y_true_all, y_pred_all)
        rmse = root_mean_squared_error(y_true_all, y_pred_all)
        results[name] = {"r2": r2, "rmse": rmse}
        print(f"{name}: {cfg.n_splits}-fold R^2 = {r2:.3f}, RMSE = {rmse:.3f}")

        save_model_results(
            model=name, label=run, gp=gp, feature_names=X.columns.tolist(),
            n_transformed_features_min=min(n_feat_all), n_transformed_features_max=max(n_feat_all),
            n_transformed_features_last=n_feat_all[-1], n_fit_rows=n_fit_rows,
            y_true=y_true_all, y_pred=y_pred_all, y_std=y_std_all, fit_seconds=fit_seconds,
            cfg=cfg, out_dir=out_dir,
        )
        # save_model_object(model=name, label=run, gp=gp, pre=pre, out_dir=out_dir)

    print(f"\nSummary ({cfg.n_splits}-fold):", {k: {m: round(v, 3) for m, v in d.items()} for k, d in results.items()})
    return results



def run_traintest(cfg):
    """cfg.test_size holdout split (default 80:20)."""
    results = {}
    run = f"traintest_{round((1 - cfg.test_size) * 100)}_{round(cfg.test_size * 100)}"
    out_dir = _new_run_dir(cfg, run) if cfg.save_results else None

    for name, df in cfg.dataframes.items():
        pre, X, y, _ = build_Xy(df, name, cfg)
        Xtr, Xte, ytr, yte = train_test_split(
            X, y, test_size=cfg.test_size, random_state=cfg.split_random_state
        )

        Xtr = pre.fit_transform(Xtr); Xte = pre.transform(Xte)
        gp = make_gp(Xtr.shape[1], cfg)
        t0 = time.perf_counter()
        gp.fit(Xtr, ytr)
        fit_seconds = time.perf_counter() - t0
        y_pred, y_std = gp.predict(Xte, return_std=True)    #instead of passing the gp make a pipe line to pass the xy transfermer for proper scaling of features 

        r2 = r2_score(yte, y_pred)
        rmse = root_mean_squared_error(yte, y_pred)
        results[name] = {"r2": r2, "rmse": rmse}
        print(f"{name}: {run} R^2 = {r2:.3f}, RMSE = {rmse:.3f}")

        save_model_results(
            model=name, label=run, gp=gp, feature_names=X.columns.tolist(),
            n_transformed_features_min=Xtr.shape[1], n_transformed_features_max=Xtr.shape[1],
            n_transformed_features_last=Xtr.shape[1], n_fit_rows=Xtr.shape[0],
            y_true=yte, y_pred=y_pred, y_std=y_std, fit_seconds=fit_seconds, cfg=cfg, out_dir=out_dir,
        )
        # save_model_object(model=name, label=run, gp=gp, pre=pre, out_dir=out_dir)

    print(f"\nSummary ({run}):", {k: {m: round(v, 3) for m, v in d.items()} for k, d in results.items()})
    return results
```
